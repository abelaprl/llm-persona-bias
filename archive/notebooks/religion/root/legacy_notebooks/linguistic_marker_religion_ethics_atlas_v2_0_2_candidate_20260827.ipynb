## Jelajah hasil berpasangan

```python
within = pd.read_csv(TABLES / 'paired_within_identity_all_scales.csv.gz')
direct = pd.read_csv(TABLES / 'paired_direct_identity_all_scales.csv.gz')
display(within.groupby(['population','multiplicity_family','outcome_tier']).size().rename('rows'))
display(direct.groupby(['population','outcome_tier']).size().rename('rows'))
```

```python
def query_within(model=None, subset=None, identity=None, tier=None, significant_only=False):
    result = within.copy()
    for column, value in [('model_id',model),('subset',subset),('identity',identity),('outcome_tier',tier)]:
        if value is not None:
            result = result[result[column] == value]
    if significant_only:
        result = result[result.p_holm < 0.05]
    return result.sort_values(['p_holm','outcome_id'])

display(query_within(subset='commonsense', tier='core_confirmatory', significant_only=True).head(50))
```
