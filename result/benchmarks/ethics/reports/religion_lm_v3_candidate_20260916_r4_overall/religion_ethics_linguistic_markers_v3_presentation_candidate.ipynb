# Religion ETHICS — linguistic marker kandidat v3

Pool seluruh empat subset; enam model dan 17 persona. Tidak ada akurasi ETHICS gabungan. Marker final-only mencakup abstain berteks. Status kandidat: audit dua reviewer dan recall independen belum selesai.

Buka `manifest.json`, `registry/markers.json` serta tabel setiap gambar untuk denominator dan provenance. M/C/J/A = Muslim/Christian/Jewish/Atheist; E = explicit; I1–I3 = implicit; N = neutral.

Pada forest plot, baris tanpa titik berarti efek tidak tersedia; titik tanpa garis CI berarti interval tidak dapat diestimasi. Perhatikan n pasangan. Efek text-conditional hanya berlaku pada pasangan dengan teks di kedua sisi; lihat output-level dan availability untuk konteks seleksi.


[Penjelajah seluruh 52 kontras, termasuk I1/I2/I3](evidence/all_contrasts_explorer.html). Filter model, marker, populasi dan metrik; tabel menyertakan jumlah occurrence mentah kedua sisi. CI 95% bersifat marginal; koreksi Holm dilaporkan terpisah.


Istilah penjelasan pada grafik mengikuti kategori operasional linguistic_text: final dengan kata linguistik, termasuk verdict leksikal atau prosa pendek. Kategori ini tidak membuktikan bahwa model memberikan argumentasi panjang. Label-only merujuk pada format verdict biner teknis yang didefinisikan surface.

Prevalence tanpa pembobotan memakai McNemar eksak. Uji density dan prevalence equal-subset memakai paired sign-flip (eksak hingga 16 pasangan informatif; selebihnya 10.000 pengacakan). Asumsinya pertukaran kondisi dalam pertanyaan sah di bawah hipotesis nol. CI bootstrap tetap perkiraan; interval degenerat, kurang dari enam pasangan informatif, atau dukungan strata tidak cukup ditandai NA. Holm mempertahankan seluruh hipotesis yang direncanakan. ## Audit dan keterbatasan

[Bukti lengkap dengan highlight](evidence/full_responses.html). Tema leksikal tidak sama dengan dukungan model, truth doktrinal atau moral judgement. Scope yang tidak pasti tetap unresolved. Dua putaran discovery masih menemukan gap; tidak ada klaim coverage hampir sempurna. Sensitivitas tidak memperbaiki cacat upstream pada prompt/scoring.
