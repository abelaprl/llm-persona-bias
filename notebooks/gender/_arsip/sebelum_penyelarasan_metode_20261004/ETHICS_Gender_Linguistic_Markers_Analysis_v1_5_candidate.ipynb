Perubahan label dan kontras dilakukan **langsung di notebook ini**. Sel berikut merender PNG dari tabel beku, memeriksa hash, dan mencadangkan gambar aktif sebelum menggantinya.
```python
from pathlib import Path
from datetime import datetime
import hashlib, html, json, re, shutil, sys
import pandas as pd
from matplotlib.figure import Figure
from matplotlib.text import Text
from matplotlib.collections import PathCollection
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "src/MAIN_PARSING/ETHICS").is_dir())
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
RUN_TAG = datetime.now().strftime("%Y%m%d_%H%M%S_%f")
def file_hash(path):
    return hashlib.sha256(Path(path).read_bytes()).hexdigest()

def gender_display(text):
    text = re.sub(r"(?<![\w-])(?:Laki-laki|laki-laki)(?![\w-])", "Man", str(text))
    text = re.sub(r"(?<![\w-])(?:Perempuan|perempuan)(?![\w-])", "Woman", text)
    text = text.replace("L·", "M·").replace("P·", "W·")
    text = text.replace("L = Man; P = Woman", "M = Man; W = Woman")
    return text

# Runtime overrides are local to this notebook kernel; source modules stay unchanged.
from src.MAIN_PARSING.ETHICS import gender_lm_terminology as terminology
terminology.REPLACEMENTS.pop("Man", None)
terminology.REPLACEMENTS.pop("Woman", None)
terminology.REPLACEMENTS.update({"Laki-laki": "Man", "laki-laki": "Man",
                                 "Perempuan": "Woman", "perempuan": "Woman"})

def presentation_save(fig, filename, *args, **kwargs):
    if Path(filename).suffix.lower() == ".svg":
        return None
    for ax in fig.axes:
        for axis_name in ("x", "y"):
            old_labels = [t.get_text() for t in getattr(ax, "get_" + axis_name + "ticklabels")()]
            new_labels = [gender_display(t) for t in old_labels]
            if old_labels != new_labels:
                getattr(ax, "set_" + axis_name + "ticks")(
                    getattr(ax, "get_" + axis_name + "ticks")(), new_labels)
    for artist in fig.findobj(Text):
        artist.set_text(gender_display(artist.get_text()))
    for ax in fig.axes:
        for collection in ax.collections:
            if isinstance(collection, PathCollection):
                if hasattr(collection.get_alpha(), '__len__'):
                    collection.set_alpha([1.0] * len(collection.get_offsets()))
                else:
                    collection.set_alpha(1.0)
        for line in ax.lines:
            if line.get_marker() not in (None, "", "None", "none"):
                line.set_alpha(1.0)
    return original_savefig(fig, filename, *args, **kwargs)

def install_presentation(bundle, active, image_folder, figure_column):
    new_index = pd.read_csv(bundle / "figure_index.csv")
    for column in ("title", "caption", "description", "page_scope"):
        if column in new_index.columns:
            new_index[column] = new_index[column].map(lambda value: gender_display(value) if pd.notna(value) else value)
    new_index.to_csv(bundle / "figure_index.csv", index=False)
    backup = active / "_arsip" / ("sebelum_man_woman_" + RUN_TAG)
    backup.mkdir(parents=True, exist_ok=False)
    for name in ("figures", "figure_index.csv", "manifest.json", "presentation_source.json",
                 "gallery.html", "README.md"):
        source = active / name
        if source.is_dir():
            shutil.copytree(source, backup / name)
        elif source.exists():
            shutil.copy2(source, backup / name)
    figures = active / "figures"
    figures.mkdir(exist_ok=True)
    for row in new_index.to_dict("records"):
        name = row[figure_column]
        name = name if name.endswith(".png") else name + ".png"
        source = bundle / image_folder / name
        assert file_hash(source) == row["png_sha256"]
        shutil.copy2(source, figures / name)
        assert file_hash(figures / name) == row["png_sha256"]
        csv = source.with_suffix(".csv")
        if csv.exists():
            assert file_hash(csv) == row["csv_sha256"]
            shutil.copy2(csv, figures / csv.name)
    expected = {str(value) if str(value).endswith(".png") else str(value) + ".png"
                for value in new_index[figure_column]}
    archived = []
    for path in list(figures.glob("*.png")) + list(figures.glob("*.svg")):
        if path.name not in expected:
            assert file_hash(path) == file_hash(backup / "figures" / path.name)
            path.unlink()  # A byte-identical, verified copy remains in the dated archive.
            archived.append(path.name)
    shutil.copy2(bundle / "figure_index.csv", active / "figure_index.csv")
    shutil.copy2(bundle / "manifest.json", active / "manifest.json")
    provenance_path = active / "presentation_source.json"
    provenance = json.loads(provenance_path.read_text(encoding="utf-8")) if provenance_path.exists() else {}
    provenance.update(print_bundle=str(bundle.relative_to(ROOT)).replace(chr(92), "/"),
                      presentation_bundle=str(bundle.relative_to(ROOT)).replace(chr(92), "/"),
                      notebook_rendered=True, display_gender="Man/Woman", direct_effect="Man - Woman",
                      image_formats=["png"], previous_presentation=str(backup),
                      style_reference="notebooks/gender/linguistic marker/GSM8K/linguistic_marker_gender_gsm8k_direct_gender_v3.ipynb")
    provenance_path.write_text(json.dumps(provenance, ensure_ascii=False, indent=2), encoding="utf-8")
    sections = []
    for row in new_index.to_dict("records"):
        name = str(row[figure_column])
        name = name if name.endswith(".png") else name + ".png"
        caption = gender_display(row.get("caption", row.get("description", name)))
        sections.append(f'<section><h2>{html.escape(name)}</h2><img src="figures/{name}"><p>{html.escape(caption)}</p></section>')
    (active / "gallery.html").write_text('<!doctype html><meta charset="utf-8"><title>Gender: Man/Woman</title><style>body{font:16px Arial;margin:2rem}img{max-width:100%;width:800px}section{margin-bottom:3rem}</style>' + "".join(sections), encoding="utf-8")
    (active / "README.md").write_text("# Visualisasi gender: Man/Woman\n\nJalankan seluruh notebook di folder ini untuk merender ulang dari tabel beku. Konfigurasi presentasi ada langsung pada sel awal notebook. PNG aktif ada di figures/ dan gallery.html; gambar lama tersedia di _arsip/.\n\nArah langsung Man − Woman; NA tetap ditampilkan, bukan nol. Warna titik penuh; signifikansi langsung dibedakan lewat simbol. Skala heatmap tetap kuantitatif. Tidak ada scoring atau uji inferensial baru. Status audit sumber tidak berubah.\n", encoding="utf-8")
    print(f"Terpasang: {len(expected)} PNG; {len(archived)} gambar lama dipindahkan ke arsip {backup}")

from src.MAIN_PARSING.ETHICS import build_gender_ethics_lm_print as renderer
renderer.TITLES = {key: gender_display(value) for key, value in renderer.TITLES.items()}
PRESENTATION_BUNDLE = ROOT / "result/ETHICS/reports" / ("gender_ethics_lm_notebook_" + RUN_TAG)
original_savefig = Figure.savefig
Figure.savefig = presentation_save
try:
    renderer.build(PRESENTATION_BUNDLE)
finally:
    Figure.savefig = original_savefig
install_presentation(PRESENTATION_BUNDLE, ROOT / "notebooks/gender/linguistic marker/ETHICS", "figures", "figure_id")
```

```python
from pathlib import Path
import hashlib, json, os, sys
import pandas as pd
from IPython.display import Image, Markdown, HTML, display

ROOT = next(x for x in [Path.cwd(), *Path.cwd().parents] if (x / 'src/MAIN_PARSING/ETHICS').is_dir())
sys.path.insert(0, str(ROOT))
BUNDLE = ROOT / 'result/ETHICS/reports/gender_lm_reference_presentation_20260920_r3'
SOURCE = ROOT / 'result/ETHICS/reports/gender_lm_v2_candidate_20260916_r3'
ACTIVE = ROOT / 'notebooks/gender/linguistic marker/ETHICS'
DOCUMENT = ACTIVE
FIGURES = DOCUMENT / 'figures'
index = pd.read_csv(ACTIVE / 'figure_index.csv')
manifest = json.loads((ACTIVE / 'manifest.json').read_text(encoding='utf-8'))
for relative, expected in manifest['source_hashes'].items():
    with (ROOT / relative).open('rb') as handle:
        actual = hashlib.file_digest(handle, 'sha256').hexdigest()
    assert actual == expected, 'Sumber frozen berubah: ' + relative

def show(figure_id):
    pages = index.loc[index.parent_figure_id.eq(figure_id)].sort_values('page')
    if pages.empty:
        raise KeyError(figure_id)
    for row in pages.itertuples():
        path = FIGURES / (row.figure_id + '.png')
        assert hashlib.sha256(path.read_bytes()).hexdigest() == row.png_sha256
        display(Image(filename=str(path), width=800))
        display(Markdown(row.caption))
        links = [f'<a href="figures/{row.figure_id}.{ext}">{ext.upper()}</a>' for ext in ['png','csv']]
        display(HTML(' · '.join(links)))

def table_link(path, label):
    return '<a href="' + os.path.relpath(path, DOCUMENT).replace(chr(92), '/') + '">' + label + '</a>'

display(HTML('<style>.jp-Notebook { max-width: none; } .output_png img { max-width: 100%; }</style>'))
```

## 7. Penjelajah lengkap: seluruh 41 marker dan kontras

Ubah `marker_id`, `scope`, dan `measurement` untuk membaca profil/kontras lengkap. Sumber tetap menyediakan
jumlah mentah, density/100 kata, prevalence, effect size, CI, p Holm, dan dukungan pasangan.
Tabel default hanya membatasi tampilan, bukan analisis. Kamus mencantumkan sumber dan adaptasi buatan sendiri.
Untuk gambar tambahan, fungsi `lihat_efek()` merender pilihan di notebook tanpa menambah file di figures.

Tabel `chosen` mempertahankan arah `left − right` sumber. Untuk grafik langsung berarah Man − Woman, gunakan `lihat_efek(['stance_hedge'], comparison='direct')`.

```python
summary = pd.read_csv(SOURCE / 'marker_summary.csv', float_precision='round_trip')
comparisons = pd.read_csv(SOURCE / 'comparisons.csv', float_precision='round_trip')
registry = pd.DataFrame(json.loads((SOURCE / 'registry.json').read_text(encoding='utf-8')))
display(registry[['marker_id','registry_scope','measurement','theoretical_source','custom_adaptation']])

marker_id = 'stance_hedge'
scope = 'overall'
measurement = 'density'
display(summary.loc[summary.marker_id.eq(marker_id) & summary.scope.eq(scope)])
chosen = comparisons.loc[comparisons.marker_id.eq(marker_id) & comparisons.scope.eq(scope) & comparisons.measurement.eq(measurement)]
display(chosen[['model_id','left','right','n_pairs','n_changed','effect','measurement_unit','ci_low','ci_high','matched_rank_biserial','adjusted_p','family_id']].head(48))
display(Markdown(f'{len(chosen)} baris cocok; tampilan dibatasi 48 baris. Akses variabel `chosen` untuk semuanya.'))

def lihat_efek(marker_ids, comparison='neutral', measurement='density'):
    from src.MAIN_PARSING.ETHICS.gender_lm_reference_data import select_contrasts
    from src.MAIN_PARSING.ETHICS.gender_lm_reference_plots import effects
    import matplotlib.pyplot as plt
    if comparison == 'direct':
        from src.MAIN_PARSING.ETHICS.build_gender_ethics_lm_bab6_viz import select_direct, plot_direct_effects
        selected = select_direct(comparisons, marker_ids, measurement)
        fig = plot_direct_effects(selected, marker_ids, 'Penjelajah langsung Man − Woman', measurement)
    else:
        selected = select_contrasts(comparisons, marker_ids, comparison, measurement)
        fig = effects(selected, marker_ids, 'Penjelajah kontras terpilih', comparison, measurement)
    display(fig)
    plt.close(fig)
    return selected
```
